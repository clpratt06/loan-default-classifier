### Introduction


This project uses a random forest classifier to forecast default probability on an existing portfolio of issued loans. The primary objective is to maximize recall among predicted defaulters based on borrower characteristics and loan information, with model performance being evaluated using recall, precision, and F1-score. This might be useful in the context of loss provisioning, as we would want to understand which outstanding loans have high probability of defaulting.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report
```

```python
loan_default = pd.read_csv("Loan_Default.csv")
loan_default.head()
```

```python
loan_default.describe()
```

### Data Description

The following core variables were identified to eliminate variable redundancy and provide a balanced set of borrower and loan characteristics to help the model capture loan risk factors:

Explanatory Variables:

* loan_amount: Amount of money being borrowed.

* rate_of_interest: Interest rate of the loan.      

* term: Length of the loan in months.

* Credit_Score: Applicant's credit score.         

* LTV: Loan-to-value ratio, amount of a loan divided by the value of the asset being financed.        

* dtir1: Debt-to-income ratio, total monthly debt payments divided by gross monthly income.    

* business_or_commercial: Whether the loan is for business/commercial purposes.

Target Variable:

* Status: Whether or not the loan defaulted.

### Data Cleaning

The business_or_commercial feature indicates whether the loan is personal (nob/c) or business/commercial (b/c), and thus was converted into an indicator variable (1 = business/commercial, 0 = personal) for simplicity.

```python
loan_default['business_or_commercial_ind'] = loan_default['business_or_commercial'].map({'b/c':1, 'nob/c':0})
```

The core features identified in the Data Description were selected and features outside the core set were removed.

```python
loan_default = loan_default[['loan_amount', 'rate_of_interest', 'term', 'Credit_Score', 'LTV', 'dtir1', 'business_or_commercial_ind', 'Status']]
loan_default.head()
```

```python
loan_default.isnull().sum()
```

Variables with missing values (rate_of_interest, Credit_Score, LTV, dtir1) were imputed using the median to limit the influence of outliers.

Rows with missing values in term were dropped since the proportion of affected observations was negligible.

```python
for col in ['rate_of_interest', 'Credit_Score', 'LTV', 'dtir1']:
    loan_default[col] = loan_default[col].fillna(loan_default[col].median())

loan_default = loan_default.dropna(subset=['term'])
```

```python
loan_default.isnull().sum()
```

Inspection of the LTV variable revealed extreme outliers. Of the 148,000 observations, only six had a LTV greater than 300. Of these six, all were greater than 2000. Thus, these extreme outliers were removed to preserve the integrity of the analysis.

```python
loan_default[loan_default['LTV'] > 300]
```

```python
loan_default = loan_default[loan_default['LTV'] <= 300]
```

### Modeling and Interpretation

The data was split into testing and training sets, stratifying status to ensure the proportion of defaulting loans was the same in estimation and evaluation. The testing and training sets were split into the explanatory and target variables, before fitting a random forest classifier to the training set. We use balanced class weights to address class imbalance, as the majority of loans did not default. (See EDA) Even though defaults are the minority class, they are the most important to identify in the context of loans.

```python
train, test = train_test_split(loan_default, test_size=0.2, stratify=loan_default['Status'])

x_train = train[['loan_amount', 'rate_of_interest', 'term', 'Credit_Score', 'LTV', 'dtir1', 'business_or_commercial_ind']]
y_train = train['Status']

x_test = test[['loan_amount', 'rate_of_interest', 'term', 'Credit_Score', 'LTV', 'dtir1', 'business_or_commercial_ind']]
y_test = test['Status']

model = RandomForestClassifier(class_weight='balanced', n_estimators=100, random_state=42)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict_proba(x_test)[:, 1]
y_pred = (y_pred >= 0.5).astype(int)
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
print(classification_report(y_test, y_pred))
```

The random forest classifier achieved a recall of 0.93 for defaulting loans, meaning 93% of actual defaults were correctly identified. However, precision is 0.77, indicating that many of the predicted defaults were false positives. Because failing to identify a defaulting borrower is more costly than incorrectly flagging a non-defaulting one, prioritizing recall is arguably economically justified in this context. That being said, an F1-score of 0.84 for defaulters and accuracy of 0.91 indicates strong model performance across the board.

```python
feature_importance_df = pd.DataFrame({
    "feature": model.feature_names_in_,
    "importance": model.feature_importances_
}).sort_values(by="importance", ascending=False)
feature_importance_df
```

### Conclusion

In this project, we utilized a random forest classifier to predict loan defaults, achieving a recall of 0.93 for defaulting loans. While the precision is lower at 0.77, it is important to note that falsely flagging loan defaults has a drastically lower operational cost than failing to identify them. Thus, recall is prioritized over precision in this setting.
